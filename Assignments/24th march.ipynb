```python
from sklearn.feature_extraction.text import TfidfVectorizer

# Step 1: Documents
documents = [
    "Machine learning is a part of artificial intelligence",
    "Artificial intelligence and machine learning are related fields",
    "Deep learning is a subset of machine learning",
    "Data science uses machine learning and statistics",
    "Python is widely used in data science and machine learning"
]

# Step 2: TF-IDF Vectorizer
vectorizer = TfidfVectorizer(stop_words='english')

tfidf_matrix = vectorizer.fit_transform(documents)

# Step 3: Get feature names
feature_names = vectorizer.get_feature_names_out()

# Step 4: Extract top keywords for each document
for i, doc in enumerate(tfidf_matrix):
    print(f"\nDocument {i+1} Top Words:")
    sorted_words = doc.toarray().flatten().argsort()[::-1]
    
    for idx in sorted_words[:5]:  # top 5 words
        print(feature_names[idx], ":", round(doc.toarray()[0][idx], 3))
```

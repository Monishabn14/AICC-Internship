```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

# Step 1: Training Data (simple dataset)
train_reviews = [
    "amazing movie",
    "great acting",
    "loved the film",
    "fantastic story",
    "bad movie",
    "worst film",
    "boring and dull",
    "waste of time"
]

train_labels = [1, 1, 1, 1, 0, 0, 0, 0]  # 1 = Positive, 0 = Negative

# Step 2: Convert text to numbers
vectorizer = CountVectorizer()
X_train = vectorizer.fit_transform(train_reviews)

# Step 3: Train model
model = MultinomialNB()
model.fit(X_train, train_labels)

# Step 4: Test Reviews
test_reviews = [
    "The movie was amazing with great acting and story",
    "I really loved the cinematography and direction",
    "The film was boring and too long",
    "Worst movie ever, waste of time",
    "It was okay, not the best but enjoyable"
]

X_test = vectorizer.transform(test_reviews)

# Step 5: Predictions
predictions = model.predict(X_test)

# Step 6: Display Results
for review, pred in zip(test_reviews, predictions):
    sentiment = "Positive" if pred == 1 else "Negative"
    print(f"Review: {review}")
    print(f"Sentiment: {sentiment}\n")
```
